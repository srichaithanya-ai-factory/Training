```python
mixed_list = [10, "hello", 3.14, 42, "world", 2.718, 7]

counts = {"int": 0, "str": 0, "float": 0}
for item in mixed_list:
    if type(item) == int:
        counts["int"] += 1
    elif type(item) == str:
        counts["str"] += 1
    elif type(item) == float:
        counts["float"] += 1

print(counts)
```

```python
weight_kg = 70
height_m = 1.75

bmi = weight_kg / (height_m ** 2)

if bmi < 18.5:
    category = "Underweight"
elif bmi < 25:
    category = "Normal"
elif bmi < 30:
    category = "Overweight"
else:
    category = "Obese"

print(f"BMI: {bmi:.1f} - {category}")
```

```python
target = 7
guesses = [2, 9, 7]  # Simulated user inputs
idx = 0

while idx < len(guesses):
    guess = guesses[idx]
    print(f"Guessed: {guess}")
    if guess == target:
        print("Correct!")
        break
    idx += 1
```

```python
def calculate(*args, multiply=False):
    if not args:
        return 0
    if multiply:
        result = 1
        for num in args:
            result *= num
        return result
    return sum(args)

print(calculate(2, 3, 4))                  
print(calculate(2, 3, 4, multiply=True)) 
```

```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Cannot divide by zero!"
    except TypeError:
        return "Error: Please provide numbers only!"

print(safe_divide(10, 0))
print(safe_divide(10, "two"))
```

```python
dirty_data = ["10", "20x", "30", "abc", "50"]

def clean_item(item):
    try:
        return int(item)
    except ValueError:
        return None


clean_data = [clean_item(x) for x in dirty_data if clean_item(x) is not None]
print(clean_data)  # Expected: [10, 30, 50]
```
