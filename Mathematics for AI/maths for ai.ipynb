##Matrix Multiplication

```python
import numpy as np
A = np.array([
    [1, 2],
    [3, 4]
])
B = np.array([
    [5, 6],
    [7, 8]
])
result = np.matmul(A, B)
print(result)
```

```python
#Multiply a Matrix by a Vector
import numpy as np
A = np.array([
    [1,2],
    [3,4]
])
v = np.array([5,6])
result = A @ v
print(result)
```

```python
#Check Matrix Shapes Before Multiply
import numpy as np
A = np.array([
    [1,2],
    [3,4]
])
B = np.array([
    [5,6],
    [7,8]
])
print("Shape of A:", A.shape)
print("Shape of B:", B.shape)
result = A @ B
print(result)
```
