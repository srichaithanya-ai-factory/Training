4. Linked List

Unlike arrays, linked lists store nodes.

Each node contains

Data
Next pointer

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


head = Node(10)
head.next = Node(20)
head.next.next = Node(30)

temp = head

while temp:
    print(temp.data)
    temp = temp.next
```

```python
from collections import deque

queue = deque()

queue.append(10)

queue.append(20)

queue.append(30)

print(queue.popleft())
```

Binary Search

Works only on sorted arrays.

1 3 5 7 9 11

Repeatedly divide by half.

```python
def binary_search(arr,target):

    low = 0
    high = len(arr)-1

    while low<=high:

        mid=(low+high)//2

        if arr[mid]==target:
            return mid

        elif arr[mid]<target:
            low=mid+1

        else:
            high=mid-1

    return -1


arr=[2,5,8,11,15]

print(binary_search(arr,11))
```
